## Dictionaries

```python
## Creating a dictionaries

dict1={"Krishanu":2023,"Akash":2021,"Roshan":2022}
```

```python
print(dict1)
```

```python
print(type(dict1))
```

```python
for i in dict1:
    print(i)
```

```python
for i in dict1.values():
    print(i)
```

```python
for i in dict1.items():
    print(i)
    
```

```python
car1_model={"Mercedes":1960}
car2_model={"Audi":1970}
car3_model={"Ambassador":1968}

car_type={'car1':car1_model,'car2':car2_model,'car3':car3_model}
```

```python
print(car_type)
```

```python
print(car_type['car1'])
```

```python
print(car_type['car1']['Mercedes'])
```
